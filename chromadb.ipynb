```python
import chromadb
import ollama

# Must match the exact model you used in your indexer
EMBEDDING_MODEL = "hf.co/CompendiumLabs/bge-base-en-v1.5-gguf"

# 1. Initialize client and get collection (Do NOT pass an embedding_function)
client = chromadb.PersistentClient(path="./chroma_db")
collection = client.get_collection(name="srd_docs")

# 2. Define your search query text
query_text = "How does a fireball spell work?"

# 3. Embed your search query using the exact same Ollama model
response = ollama.embed(model=EMBEDDING_MODEL, input=query_text)
query_embedding = response["embeddings"][0]

# 4. Query using query_embeddings instead of query_texts
results = collection.query(query_embeddings=[query_embedding], n_results=5)

# doc is a list of documents, must loop through doc as well
for i, doc in enumerate(results['documents'][0]):
    print(f"Document {i + 1}: {doc}")
```
